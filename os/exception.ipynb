```python
class InvalidAgeError(Exception):
    pass
try:
    age=19
    if age<18:
        raise InvalidAgeError("you are not eligible to vote")
    print("you can vote")
except InvalidAgeError as e:
    print(e)        
```

```python
class InvalidBalance(Exception):
    pass
try:
    bal=5000
    withdrawal=float(input("enter amount:"))
    if withdrawal>bal:
        raise InvalidBalance("insufficient balance")
    print("you can withdraw")
except InvalidBalance as e:
    print(e)

```

| Mode  | Meaning                 |
| ----- | ----------------------- |
| `"r"` | Read (default)          |
| `"w"` | Write (overwrites file) |
| `"a"` | Append                  |
| `"x"` | Create new file         |



* READ MODE(r)

```python
file=open("poit.txt","r")
```

```python
file.read() 
```

```python
file.readline()
```

```python
file.readlines()
```

```python
file.close()
```

```python
with open('poit.txt')as file:
    print(file.read())
```

* WRITE MODE(w)

```python
file=open('poit.txt','w') ##to writw in a file
file.write("hluuuuuuu!")
file.close()
file=open('poit.txt','r')
file.read()
```

```python
file=open('poit1.txt','w') #file if didnt exist,it creat a file  
file.write("hluuuuuuu!")
file.close()
file=open('poit.txt','r')
file.read()
```

* APPEND MODE(a)

```python
file=open('poit.txt','a')
file.write("File handling in Python means creating, reading, writing, appending, and managing files using built-in functions like open().")
file.close()
```

```python
open('poit.txt','r').read()
```

```python
file=open('filehandling.txt','a') ##
file.write("File handling in Python means creating, reading, writing, appending, and managing files using built-in functions like open().")

file.close()
```

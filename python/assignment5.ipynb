que1. Create a class Employee with attributes emp_name, emp_id, and salary. Create 3 objects and display their details.

```python
class Employee:
    def __init__(self,emp_name,emp_id,salary):
        self.emp_name=emp_name
        self.emp_id=emp_id
        self.salary=salary
    def display(self):
        print("employee name:",self.emp_name)
        print("employee id:",self.emp_id)
        print("employee salary:",self.salary)
# creating three objects        
obj1=Employee("nisha",234,20000) 
obj2=Employee("harshita",239,70000) 
obj3=Employee("diya",456,90000)
#method call
obj1.display()
obj2.display()
obj3.display()  
#OUTPUT
```

que2.Create a class Calculator with methods for addition, subtraction, multiplication, and division.

```python
#CODE
class Calculator:
    def addition(self):
        a=float(input("enter your number:"))
        b=float(input("enter yor number:"))
        self.s=a+b
        print("sum of given numbers:",self.s)
    def subtraction(self):
        c=float(input("enter your number:"))
        d=float(input("enter your number:"))
        self.dif=c-d
        print("difference of given numbers:",self.dif)
    def multiplication(self):
        e=float(input("enter your number:"))
        f=float(input("enter your number:"))
        self.m=e*f
        print("product of given numbers:",self.m)
    def divison(self):
        g=float(input("enter your number:"))
        h=float(input("enter your number:"))
        self.r=g%h
        print("remainder is:",self.r)
#object defining        
obj=Calculator() 
#method call
obj.subtraction()
obj.addition()  
obj.multiplication()
obj.divison()  
#OUTPUT
```

que3.Create a class Book with a constructor to initialize title, author, and price, then print details. 

```python
#CODE
class Book:
    def __init__(self,title,author_name,price):
        self.title=title
        self.author_name=author_name
        self.price=price
    def display(self):
        print(f"The title of book is {self.title} ") 
        print("Author of book:",self.author_name)  
        print("price of book:",self.price) 
#defining object        
obj=Book("gillu","mahadevi verma",50)
#method call
obj.display()
#OUTPUT
```

que4.Create a class Vehicle with method display(). Inherit it into class Car and call parent method. 

```python
#CODE
class Vechile:
    def display(self):
        print("This is a vechile")
class Car(Vechile):
    def show(self):
        self.display()
        print("this is car")
#defining object        
obj=Car()
#method call
obj.show()
#OUTPUT
```

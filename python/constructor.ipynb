## Constructor in Python

A constructor in Python is a special method that is automatically called whenever an object of a class is created.

Its main purpose is to initialize the object by assigning initial values to its attributes or performing setup operations.

---

## Purpose of a Constructor

* Initializes object data
* Sets default or user-provided values
* Makes the object ready to use immediately after creation

---

## Special Constructor Method

Python uses a special method called `__init__()` as the constructor.

This method runs automatically when an object is created.

---

## Types of Constructors

### 1. Default Constructor

A constructor that does not take additional arguments.
It initializes objects with default values.

### 2. Parameterized Constructor

A constructor that accepts arguments.
It allows different objects to have different initial values.

---

## Role of `self`

`self` refers to the current object of the class.
It is used to access and store the data related to that specific object.

---

## Real-Life Example

A constructor is similar to filling out details when creating a new account in a bank:

* Name
* Account number
* Initial balance

As soon as the account is created, all these details are initialized automatically.

---

## Advantages of Constructors

* Automatic initialization of objects
* Reduces repetitive work
* Improves code organization
* Makes programs easier to manage

---

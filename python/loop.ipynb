### looping statement in python
~ A loop in Python is a control flow statement used to repeatedly execute a block of code until a specific condition is met or all items in a sequence are processed. Loops automate repetitive tasks, saving you from writing the same code multiple times.

1. for loop 
2. while loop
#### for loop:
The Python for loop iterates over a sequence (list, tuple, dictionary, set, or string) and executes the code block once for each item.
#### syntax:
for x in variable_name:

      print(x)
#### while loop:
A Python while loop repeatedly executes its code block as long as a given boolean condition remains True






```python
### examples of for loop
name="ritik"
for x in name:
    print(x) 
```

```python
a=[1,2,3,4,5]
for y in a:
    print(y)
```

```python
##use of range function
for x in range (1,21):
    print(x)
```

```python
for x in range(1,20,2): #(start,end,step),end-1 print hoga
    print(x)
    
```

```python
dict={"name":"swati",
     "class":2,"address":"jaipur"}
print(dict) 
print(">>>>>keys of dict:")
for x in dict.keys():
    print(x) 
print(">>>>>>>values of dict:")    
for x in dict.values():
    print(x)  
print(">>>>>items of dict:")
for x in dict.items():
    print(x)         
```

```python
tpl=(1,2,34,5,6,7,67) #loop in tuple
for x in tpl:
    print(x)
```

```python
lst=[1,2,3,4,5,6] #use of break keyword
for x in lst:
    if x==5:
        break #used for stopping given loop at any given condition
    print(x)
```

```python
fruit=["apple","mango","orange","kiwi","chikoo"] #use of keyword continue
for x in fruit:
    if x=="mango":
        continue #use for skipping an item
    print(x)    
```

```python
lst=[["apple",1,2,3],"hello","hi"] #nested list
print(lst)
print(lst[0][2]) #indexing
```

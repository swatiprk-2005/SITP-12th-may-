### inheritance:
Inheritance in Python is an Object-Oriented Programming (OOP) feature where one class acquires the properties and methods of another class.

The existing class is called the Parent/Base/Super class
The new class is called the Child/Derived/Sub class

#### Inheritance helps with:

* Code reuse
* Better organization
* Extending existing functionality

| Type of Inheritance          | Definition                                            | Structure                      | Example                                |
| ---------------------------- | ----------------------------------------------------- | ------------------------------ | -------------------------------------- |
| **Single Inheritance**       | One child class inherits from one parent class        | `Parent → Child`               | `class Child(Parent)`                  |
| **Multiple Inheritance**     | One child class inherits from multiple parent classes | `Parent1 + Parent2 → Child`    | `class Child(Father, Mother)`          |
| **Multilevel Inheritance**   | A child class becomes parent for another class        | `Grandparent → Parent → Child` | `class Child(Parent)`                  |
| **Hierarchical Inheritance** | Multiple child classes inherit from one parent class  | `Parent → Child1, Child2`      | `class Child1(Parent)`                 |
| **Hybrid Inheritance**       | Combination of two or more inheritance types          | Mixed structure                | Combination of multiple + hierarchical |

* summary table

| Type         | Description                      |
| ------------ | -------------------------------- |
| Single       | One parent → one child           |
| Multiple     | Multiple parents → one child     |
| Multilevel   | Chain of inheritance             |
| Hierarchical | One parent → many children       |
| Hybrid       | Combination of inheritance types |

```python
class Person:
    def person_info(self):
        print("im a person")
class Teacher(Person):        
    def teacher_info(self):
        print("im a teacher")
class Hod(Teacher):
    def hod_info(self):
        print("im a hod")
a=Hod()
print(a.person_info())
print(a.teacher_info())
print(a.hod_info())

```

```python
#### MULTIPLE INHERITANCE:
class Developer():
    def developer_info(self):
        print("developer data")
class Manager():
    def manager_info(self):
        print("manager data")
class Teamleader(Developer,Manager):
    def teamleader_info(self):
        print("teamleader info")  
a=Teamleader()
print(a.developer_info())
print(a.manager_info())
print(a.teamleader_info())
```

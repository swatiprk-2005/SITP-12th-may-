```python
class ATM:
    def __init__(self):
        self.d=int(input("enter depoist amount:"))
        self.w=int(input("enter withdrwal amount:"))
        self.balance=20000
    def depoist(self):
        print("your depoisted money:",self.d)
    def withdraw(self):
        print("your withdrwal amount:",self.w)
    def balance1(self):
        if self.w>0:
            if self.d>0:
                self.balance=self.balance+self.d-self.w 
            else:
                self.balance=self.balance-self.w   
        else:
            self.balance=self.balance+self.d
        print("your balance",self.balance)
a=ATM()
a.depoist()
a.withdraw()     
a.balance1()       

```

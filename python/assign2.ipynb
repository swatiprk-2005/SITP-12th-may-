* Que2.Create a class PrimeChecker with a function to check whether a number is prime or not. 

```python
class PrimeChecker :
    def prime(self):
        for x in range(1,11):
            if x>1:
                for i in range(2,x):
                    if x%i==0:
                        break
                else:
                    print("prime",x)
a=PrimeChecker()        
a.prime()            
```

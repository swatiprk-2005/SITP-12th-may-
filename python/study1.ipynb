que1. give a number n  ,print table from 1 to 10 for n in a single line?

```python
a=int(input("enter your number:"))
for i in range(1,11):
    b=a*i
    print(b ,end=" ")
    i+=1
```

que3. given a number x,the task is to print the numbers from x to 0 in decreasing order in single line

```python
x=int(input("enter your number:"))
while x>=0:
    print(x,end=" ")
    x-=1
```

que4. given a positive integer x,the task is to print the numbers from 1 to x in the order as 1**2, 2**2.........

```python
x=int(input("enter your number:"))
for x in range(1,x+1):
    a=x**2
    print(a,end=" ")
```

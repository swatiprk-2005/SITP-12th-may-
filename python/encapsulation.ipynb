## Encapsulation in Python

Encapsulation is an Object-Oriented Programming (OOP) concept that means **combining data and the methods that operate on that data into a single unit (class)** and **restricting direct access to some parts of the object**.

In simple words, encapsulation is used to **hide the internal details of an object and allow access only through controlled methods**.

---

## Main Idea

A class keeps:

* **Data (attributes/variables)**
* **Functions (methods)**

together in one place and protects the data from unwanted access or modification.

---

## Why Encapsulation is Important

* Protects data from accidental changes
* Provides security
* Makes programs easier to maintain
* Hides implementation details
* Allows controlled access to data

---

## Access Levels in Python

Python uses naming conventions to indicate access levels:

### 1. Public Members

Accessible from anywhere.

### 2. Protected Members

Intended for internal use within the class and subclasses.

### 3. Private Members

Restricted from direct access outside the class using name mangling.

---

## How Encapsulation Works

Instead of allowing direct access to data, the class provides controlled ways to:

* Read data
* Update data
* Validate changes

This ensures that object data remains safe and consistent.

---

## Real-Life Example

Think of an ATM machine:

* You cannot directly access the bank database.
* You interact through buttons and options.
* The internal process is hidden from the user.

This is encapsulation:

* Data is hidden
* Access is controlled through methods

---

## Advantages of Encapsulation

* Data hiding
* Better security
* Improved flexibility
* Easier maintenance
* Better code organization

---

## Key Point

Encapsulation in Python means:

> **Hiding internal data and providing controlled access to it through class methods.**


```python
class College:
    college_name="arya" #public variable
    college_garden="yes"#public variable
    college_account="2345678890"#public variable
a=College()
a.college_name
a.college_garden
a.college_account

```

```python
class College:
    college_name="arya" #public variable
    college_garden="yes"#public variable
    __college_account="2345678890"#private variable
a=College()
a.__college_account
```

```python
class College:
    college_name="arya" #public variable
    _college_garden="yes"#protected variable
    __college_account="2345678890"#private variable
a=College()
a._college_garden
```

```python
a._college_garden
```
